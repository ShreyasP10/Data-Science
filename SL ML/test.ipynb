```python
from sklearn.linear_model import LogisticRegression


X = [[1],[2],[3],[4],[5],[6]]

y=[0,0,0,1,1,1]

model =LogisticRegression()
model.fit(X,y)
hours=1
result=model.predict([[hours]])[0]

if result==1:
    print(f"Based on hours {hours}, you are likely to PASS")
else:
    
    print(f"Based on hours {hours}, you are likely to FAIL")
```

```python
from sklearn.neighbors import KNeighborsClassifier

X=[
    [180,7],
    [200,7.5],
    [250,8],
    [300,8.5],
    [330,9],
    [360,9.5]
]
y=[0,0,0,1,1,1]#0:ppke , 1:Orage

model=KNeighborsClassifier(n_neighbors=3)
model.fit(X,y)

weight=180
size=8.1

prediction=model.predict([[weight,size]])[0]

if prediction==0:
    print("This is Apple")
else:
    print("This is Orange")
```

```python
from sklearn.tree import DecisionTreeClassifier 
X=[
    [7,2],#apple
    [8,3],#apple
    [9,8],#orange
    [10,9]#orange
]
y=[0,0,1,1]#0:apple 1:ornage

model=DecisionTreeClassifier()
model.fit(X,y)

size=8.9
shade=8.9

result=model.predict([[size,shade]])[0]

if result==0:
    print("This is Apple")
else:
    print("This is Orange")
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error

real_scores=[90,60,80,100]
pred_scores=[92,52,81,88]

mae=mean_absolute_error(real_scores,pred_scores)
mse=mean_squared_error(real_scores,pred_scores)
rmse=root_mean_squared_error(real_scores,pred_scores)

print("MAE: On Average off by ",mae)
print("MSE: Squared Mistake Value", mse)
print("RMSE: Final Realistic Error", rmse)
```

```python
# mini project

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,root_mean_squared_error
import pandas as pd
import numpy as np


data=pd.read_csv("student.csv")

X=data[['Hours']]
y=data['Score']

model=LinearRegression()
model.fit(X,y)
predicted_score=model.predict(X)


mae=mean_absolute_error(y,predicted_score)
mse=mean_squared_error(y,predicted_score)
rmse=np.sqrt(mse)

print("MAE: On Average off by ",mae)
print("MSE: Squared Mistake Value", mse)
print("RMSE: Final Realistic Error", rmse)


hours = float(input("Enter study hours per day: "))

predicted = model.predict([[hours]])
print(f"Predicted Score for {hours} hours of study: {predicted[0]}")
```
